```python
import random
options=['stone','paper','scissors']
print_h=0
print_c=0
round=int(input("how many rounds you wanna play??"))
rounds=0
while round>rounds:
    human_player=input("Enter your preference...").lower()
    print('Your response is...',human_player)
    comp=random.choice(options);
    print('The computer response is...',comp)
    if human_player==comp:
        print("It's a tie")
        point_h+=1;
        point_c+=1
    elif human_player!=comp:
        if(human_player=="stone" and comp=="scissors") or (human_player=="paper" and comp=="stone")or(human_player=="scissors" and comp=="paper"):
            point_h+=1
        else:
            point_c+=1
    rounds+=1
else:
     print("Your total score:",point_h)
     print("your computer score:",point_c)
```
