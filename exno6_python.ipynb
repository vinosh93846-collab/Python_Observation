```python
def Balance(initial_balance):
  balance = initial_balance
  def deposit(amount):
    nonlocal balance
    balance+=amount
    return balance
  return deposit

cash_1 = Balance(1000)
cash_1 = Balance(10000)
cash_2 = Balance(5000)

print(cash_1(2000))
```
