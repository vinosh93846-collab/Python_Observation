```python
from functools import reduce
prices = [500, 1200, 300, 1500, 800]

# 1. Filter products above ■500
filtered = list(filter(lambda price: price > 500, prices))

# 2. Apply 10% discount
discounted = list(map(lambda price: price * 0.90, filtered))
# 3. Calculate total bill

total = reduce(lambda x, y: x + y, discounted)
print("Original Prices:", prices)
print("Products Above ■500:", filtered)
print("After 10% Discount:", discounted)
print("Total Bill: ■", total)
```
