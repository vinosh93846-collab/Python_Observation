```python
import html
import functools

def compose(*functions):
    return lambda x: functools.reduce(lambda v, f: f(v), reversed(functions), x)

def remove_spaces(text: str) -> str:
    print("-> Running remove_spaces")
    return text.replace(" ", "")

def trim_spaces(text: str) -> str:
    print("-> Running trim_spaces")
    return text.strip()

def escape_html_tags(text: str) -> str:
    print("-> Running escape_html_tags")
    return html.escape(text)

sanitize_username = compose(
    remove_spaces,
    trim_spaces,
    escape_html_tags
)

sanitize_username("  < script >  ")
```
