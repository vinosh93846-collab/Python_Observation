```python
ph_no={0:['.',',',':'],
       1:['*','+','#'],
       2:['a','b','c'],
       3:['d','e','f'],
       4:['g','h','i'],
       5:['j','k','l'],
       6:['m','n','o'],
       7:['p','q','r','s'],
       8:['t','u','v'],
       9:['w','x','y','z']}
choice1=int(input("Enter the first number of your choice..."))
choice2=int(input("Enter the second number of your choice..."))
if choice1 in ph_no.keys() and choice2 in ph_no.keys():
    for x in ph_no[choice1]:
        for y in ph_no[choice2]:
            print(x+y)
```

```python
class Phone:

    def __init__(self):
        self.ph_no = {
            0: ['.', ',', ':'],
            1: ['*', '+', '#'],
            2: ['a', 'b', 'c'],
            3: ['d', 'e', 'f'],
            4: ['g', 'h', 'i'],
            5: ['j', 'k', 'l'],
            6: ['m', 'n', 'o'],
            7: ['p', 'q', 'r', 's'],
            8: ['t', 'u', 'v'],
            9: ['w', 'x', 'y', 'z']
        }

    def display(self):
        try:
            choice1 = int(input("Enter first number: "))
            choice2 = int(input("Enter second number: "))

            for x in self.ph_no[choice1]:
                for y in self.ph_no[choice2]:
                    print(x + y)

        except ValueError:
            print("Please enter only numbers.")

        except KeyError:
            print("Enter numbers between 0 and 9.")


obj = Phone()
obj.display()
```
