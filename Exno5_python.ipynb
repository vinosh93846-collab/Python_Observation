```python
class Payment:
    def __init__(self, base_amount):
        self.base_amount = base_amount

    def calculate_total(self):
        return self.base_amount

    def display_receipt(self):
        print(f"Base Amount: Rs. {self.base_amount:.2f}")
        print(f"Total Amount to Pay: Rs. {self.calculate_total():.2f}")

class CashPayment(Payment):
    def __init__(self, base_amount):
        super().__init__(base_amount)
        self.cash_handling_charge = 40.0

    def calculate_total(self):
        return self.base_amount + self.cash_handling_charge


class UPIPayment(Payment):
    def __init__(self, base_amount):
        super().__init__(base_amount)
        self.platform_fee = 12.0

    def calculate_total(self):
        return self.base_amount + self.platform_fee


if __name__ == "__main__":
    product_price = float(input("Enter the price of the product: "))

    print("\n--- Processing Cash Payment ---")
    cash_order = CashPayment(product_price)
    cash_order.display_receipt()

    print("\n--- Processing UPI Payment ---")
    upi_order = UPIPayment(product_price)
    upi_order.display_receipt()
```
